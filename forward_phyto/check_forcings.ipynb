```julia


N = 60
H = 6 
dz = H/N
z = collect(H:-dz:dz) .- dz/2

# Vector to store light intensity at each depth
I = zeros(N)
I2 = zeros(N)
turbidity = 0.6
I_in = 2000

for i in 1:N
    I[i] = turbidity #*corrected_z[i])
end 

I2[end] = I[end]
for i in (N-1):-1:1
    I2[i] = I[i+1] + I[i]
end

for i in 1:N
    I2[i] = I_in * exp(-I2[i] * z[i]) 
end

```

```julia

println(I2)

ll = sum(I2 .> (I_in*0.10))
println(ll)


println("depth=", z[end-ll])
```
