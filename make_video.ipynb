```python
import xarray as xr 
import matplotlib.pyplot as plt
import cmocean as cmo 
import pandas as pd 
import numpy as np
import matplotlib.dates as mdates

# Start date 
start_date = '2024-08-06'

# M = 86400
M = 8640 
end_date = pd.to_datetime(start_date) + pd.Timedelta(seconds=(10*M))
dates = pd.date_range(start_date, end_date, freq='10s')

start_date = '2024-08-06'
end_date = '2024-08-28' 

Dates = pd.date_range(start_date, end_date, freq='10s')


def format_date_ax(ax):
    ax.xaxis.set_major_locator(mdates.HourLocator(interval=4))
    date_format = mdates.DateFormatter('%m/%d %H:%M')
    ax.xaxis.set_major_formatter(date_format)


chla2cell_mc =  1e-6/0.36  # ug chl-a/ml --> pg chl-a/ml --> 0.36  pg chl-a/cell Microcystis
chla2cell_diatom = 1e-6/4 # ug chl-a/ml --> pg chl-a/ml --> 4 pg chl-a/cell Diatom
cell2chla_mc = 1/chla2cell_mc
cell2chla_diatom = 1/chla2cell_diatom
```

```python

i = 1 
adjoint = xr.open_dataset(f'backward_lambda/adjoint_{i}_august_27.nc', decode_timedelta=False)

z = adjoint["z"]
```

```python
for i in range(1, 20): 

    print(i)
    adjoint = xr.open_dataset(f'backward_lambda/adjoint_{i}_august_27.nc', decode_timedelta=False)
    fig, axs = plt.subplots(nrows=2, ncols=1, figsize=(10, 6), sharex=True, sharey=True)

    dates = Dates[adjoint.t.values]

    lambda1 = adjoint['lambda1'].values.T
    lambda2 = adjoint['lambda2'].values.T

    CLIM = 0.4
    c = axs[0].pcolormesh(dates, -z, lambda1, cmap=cmo.cm.balance , vmax=CLIM, vmin=-CLIM)#, vmin=-1, vmax=1)
    plt.colorbar(c, pad=0.01)
    plt.suptitle('Lagrangian multiplier (i=%d)' % i)
    axs[0].set_title("$\lambda_m$")

    c = axs[1].pcolormesh(dates, -z, lambda2, cmap=cmo.cm.balance , vmax=CLIM/10, vmin=-CLIM/10)#, vmin=-1, vmax=1)
    plt.colorbar(c, pad=0.01)
    axs[1].set_title("$\lambda_d$")

    axs[0].set_xlim(dates[0], dates[-1])
    format_date_ax(axs[0])

    plt.tight_layout()
    fig.savefig("gif/fig_%d.png" % i, dpi=250)
    plt.close()
```

```python
for i in range(1, 32): 
    forward = xr.open_dataset(f'forward_phyto/forward_{i}_august_27.nc', decode_timedelta=False)
    fig, axs = plt.subplots(nrows=2, ncols=1, figsize=(10, 6), sharex=True, sharey=True)

    dates = Dates[adjoint.t.values]

    lambda1 = forward['gamma1'].values.T *(86400)
    lambda2 = forward['gamma2'].values.T  *(86400)

    # ds.gamma1.values.T*(86400), cmap=cmo.cm.rain, vmin=0, vmax=1e-1)

    CLIM = 0.4
    c = axs[0].pcolormesh(dates, -z, lambda1, cmap=cmo.cm.rain, vmin=0, vmax=1e-1)#, vmin=-1, vmax=1)
    plt.colorbar(c, pad=0.01)
    plt.suptitle('Phytoplankton growth (i=%d)' % i)
    axs[0].set_title("$Microcystis$ growth")

    c = axs[1].pcolormesh(dates, -z, lambda2, cmap=cmo.cm.rain, vmin=0, vmax=1e-1)#, vmin=-1, vmax=1)
    plt.colorbar(c, pad=0.01)
    axs[1].set_title("Diatom growth")

    axs[0].set_xlim(dates[0], dates[-1])
    format_date_ax(axs[0])

    plt.tight_layout()
    fig.savefig("gif/grow_%d.png" % i, dpi=250)
    plt.close()
```
